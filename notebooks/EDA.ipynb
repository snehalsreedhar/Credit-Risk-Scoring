```python
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))

from data_prep import data_loader
from data_prep.preprocessing import preprocessor
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Checking the structure of the data
eda_df = preprocessor()
print(eda_df.dtypes)

# Checking the first five rows of the data
print(eda_df.head(5))
```

```python
# Looking at the distribution of loan amounts with a histogram
n, bins, patches = plt.hist(x=eda_df['loan_amnt'], bins='auto', color='blue',alpha=0.7, rwidth=0.85)
plt.xlabel("Loan Amount")
plt.show()
```

```python
# Plotting a scatter plot of income against age
plt.scatter(eda_df['person_income'], eda_df['person_age'],c='blue', alpha=0.5)
plt.xlabel('Personal Income')
plt.ylabel('Persone Age')
plt.show()
```

```python
# Creating a cross table of the loan intent and loan status
print(pd.crosstab(eda_df['loan_intent'], eda_df['loan_status'], margins = True))
```

```python
# Creating a cross table for home ownership and loan status and loan grade
print(pd.crosstab(eda_df['person_home_ownership'], [eda_df['loan_status'], eda_df['loan_grade']]))
```

```python
# Creating a cross table of home ownership, loan status, and average percent income
print(pd.crosstab(eda_df['person_home_ownership'], eda_df['loan_status'],
              values=eda_df['loan_percent_income'], aggfunc='mean'))
```

```python
# Creating a box plot of percentage income by loan status
eda_df.boxplot(column = ['loan_percent_income'], by = 'loan_status')
plt.title('Average Percent Income by Loan Status')
plt.suptitle('')
plt.show()
```

```python
# Creating a cross table of home ownership, loan status, and employment length
print(pd.crosstab(eda_df['loan_status'], eda_df['person_home_ownership'],
                  values=eda_df['person_emp_length'], aggfunc=['min', 'max']))
```

```python
# Creating the scatter plot for income and loan amount
plt.scatter(eda_df['person_income'], eda_df['loan_amnt'], c='blue', alpha=0.5)
plt.xlabel("Personal Income")
plt.ylabel("Loan Amount")
plt.show()
```
